# Tugas 1 – Data Preprocessing

**Tujuan:** membersihkan dan menyiapkan dataset mahasiswa sebelum digunakan dalam proses machine learning.

Tahapan yang dilakukan: load dataset, eksplorasi awal, menangani missing values, normalisasi tanggal, label encoding, split data 80:20, dan visualisasi.

## Tahap 1: Load Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# Baca dataset (pastikan file CSV berada di folder yang sama dengan notebook)
df = pd.read_csv("dataset_tugas1_preprocessing.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
```

## Tahap 2: Eksplorasi Awal (EDA)

```python
# Struktur dataset dan tipe data
df.info()

# Jumlah missing values pada setiap kolom
print("\\nJumlah nilai hilang per kolom:")
display(df.isnull().sum().to_frame(name="Jumlah Missing Values"))

# Statistik deskriptif kolom numerik dan kategorikal
display(df.describe(include="all"))
```

## Tahap 3: Menangani Missing Values

- **Nilai_Akhir:** diisi dengan modus (nilai yang paling sering muncul), karena merupakan data kategorikal berupa nilai huruf.
- **Umur:** diisi dengan median, karena berupa data numerik dan median relatif tahan terhadap nilai ekstrem.

Imputasi dilakukan sebelum encoding agar nilai kategori yang kosong dapat ditangani dengan benar.

```python
# Simpan jumlah missing values sebelum imputasi
print("Missing values sebelum imputasi:")
display(df[["Nilai_Akhir", "Umur"]].isnull().sum())

# Imputasi Nilai_Akhir dengan modus
modus_nilai = df["Nilai_Akhir"].mode()[0]
df["Nilai_Akhir"] = df["Nilai_Akhir"].fillna(modus_nilai)

# Imputasi Umur dengan median
median_umur = df["Umur"].median()
df["Umur"] = df["Umur"].fillna(median_umur)

print("Nilai modus Nilai_Akhir:", modus_nilai)
print("Nilai median Umur:", median_umur)
print("\\nMissing values setelah imputasi:")
display(df[["Nilai_Akhir", "Umur"]].isnull().sum())
```

## Tahap 4: Normalisasi Format Tanggal

Kolom `Tanggal_Ujian` dikonversi ke tipe `datetime`. Karena dataset berisi beberapa format tanggal, digunakan `format="mixed"` (tersedia pada pandas versi baru). `dayfirst=True` membantu membaca format seperti `13-04-2020` sebagai tanggal-bulan-tahun.

```python
df["Tanggal_Ujian"] = pd.to_datetime(
    df["Tanggal_Ujian"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Tipe data Tanggal_Ujian:", df["Tanggal_Ujian"].dtype)
print("Tanggal yang gagal dikonversi:", df["Tanggal_Ujian"].isnull().sum())
display(df[["Tanggal_Ujian"]].head())
```

## Tahap 5: Encoding Label

Kolom `Nama`, `Jenis_Kelamin`, `Prodi`, `Status`, dan `Nilai_Akhir` diubah dari kategori teks menjadi bilangan menggunakan `LabelEncoder`. Pemetaan kategori disimpan agar kode angka dapat ditelusuri kembali.

Catatan: `ID` tidak di-encode karena merupakan pengenal mahasiswa, bukan kategori yang diminta.

```python
kolom_kategorikal = ["Nama", "Jenis_Kelamin", "Prodi", "Status", "Nilai_Akhir"]
label_encoders = {}
mapping_kategori = {}

for kolom in kolom_kategorikal:
    encoder = LabelEncoder()
    df[kolom] = encoder.fit_transform(df[kolom].astype(str))
    label_encoders[kolom] = encoder
    mapping_kategori[kolom] = {
        kategori: int(kode)
        for kode, kategori in enumerate(encoder.classes_)
    }

print("Pemetaan kategori ke angka:")
for kolom, mapping in mapping_kategori.items():
    print(f"\\n{kolom}: {mapping}")

display(df.head())
```

## Tahap 6: Split Data (80% Data Latih, 20% Data Uji)

Karena instruksi tugas belum menentukan kolom target khusus, dataset dibagi menjadi data latih dan data uji secara langsung. Kolom `Tanggal_Ujian` diubah menjadi representasi numerik berupa jumlah hari sejak epoch agar seluruh fitur dapat diproses secara numerik.

Pembagian ini hanya contoh untuk memenuhi tahapan split. Untuk pemodelan machine learning yang sebenarnya, tentukan kolom target terlebih dahulu dan pisahkan target dari fitur.

```python
# Ubah datetime menjadi angka (hari sejak 1970-01-01)
df["Tanggal_Ujian"] = (
    df["Tanggal_Ujian"] - pd.Timestamp("1970-01-01")
).dt.days

# Antisipasi tanggal yang gagal dikonversi
df["Tanggal_Ujian"] = df["Tanggal_Ujian"].fillna(df["Tanggal_Ujian"].median())

# Pisahkan data latih dan data uji dengan proporsi 80:20
data_latih, data_uji = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("Jumlah baris data awal:", len(df))
print("Jumlah baris data latih:", len(data_latih))
print("Jumlah baris data uji:", len(data_uji))
print("Proporsi data latih: {:.0%}".format(len(data_latih) / len(df)))
print("Proporsi data uji: {:.0%}".format(len(data_uji) / len(df)))

display(data_latih.head())
display(data_uji.head())
```

## Tahap 7: Visualisasi

```python
# Histogram kolom Umur
plt.figure(figsize=(8, 5))
plt.hist(df["Umur"], bins=10, edgecolor="black")
plt.title("Histogram Umur Mahasiswa")
plt.xlabel("Umur")
plt.ylabel("Jumlah Mahasiswa")
plt.tight_layout()
plt.show()

# Grafik batang jumlah mahasiswa per Prodi.
# Gunakan data asli agar nama program studi tetap terbaca.
df_asli = pd.read_csv("dataset_tugas1_preprocessing.csv")
jumlah_per_prodi = df_asli["Prodi"].value_counts()

plt.figure(figsize=(9, 5))
jumlah_per_prodi.plot(kind="bar", edgecolor="black")
plt.title("Jumlah Mahasiswa per Program Studi")
plt.xlabel("Program Studi")
plt.ylabel("Jumlah Mahasiswa")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

## Kesimpulan

Dataset telah dieksplorasi, nilai kosong pada `Nilai_Akhir` dan `Umur` ditangani menggunakan modus dan median, tanggal dikonversi ke tipe datetime, kolom kategorikal di-encode menggunakan Label Encoding, lalu data dibagi menjadi data latih dan data uji dengan proporsi 80:20. Histogram umur dan grafik batang jumlah mahasiswa per program studi dibuat untuk membantu memahami distribusi data.
